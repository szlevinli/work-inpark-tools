# 切分 Excel

```python
import pandas as pd
from pathlib import Path
```

```python
df = pd.read_excel("/Users/lihaiqiang/Downloads/4 月退租.xlsx")
```

```python
df.info()
```

```python
grp = df.groupby("项目公司")
```

```python
branch_exp_path = Path("../output/branch")
branch_exp_path.mkdir(parents=True, exist_ok=True)
file_name_prefix = "4月退租"
```

```python
for name, group in grp:
    group.to_excel(branch_exp_path / f"{file_name_prefix}_{name}.xlsx", index=False)
```
